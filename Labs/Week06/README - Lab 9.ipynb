# Perpetuals Review

A highly relevant form of financial derivative in decentralized finance is the perpetual. Perpetuals are now the most traded crypto product: according to this [source](https://sherwood.news/crypto/perpetual-futures/) (March 2026), in **2025 alone** the ten largest centralized exchanges handled about **84 Trillion Dollars** of perpetual trading volume (CoinGecko), and decentralized perpetual exchanges handled another **7.7 Trillion Dollars** (Artemis), more than four times what they processed in the previous four years combined.

**So What is a Perpetual?**

You can think of a Perpetual as a synthetic asset, that tracks the price of an underlying asset. This synthetic asset is created by two sides of the market which are either long - expecting it to go up - or short - expecting it to go down. Within each position, market participants are basically agreeing to buy or sell the asset in the future **at the price of the asset, at the time they acquired the long or short perpetual**. Now if the price goes up, people who agreed to buy it will profit because they have an agreement to buy it for cheaper than the market price, and the same holds vice versa. Oh and one more thing, since you are only agreeing to a contract based on the asset's future price, you don't have to pay the full price of the asset. Instead you deposit a smaller amount of collateral (margin), which the protocol uses to cover your losses if the position moves against you. If you lose the value of all of your collateral below some limit defined by the protocol, your position gets liquidated, i.e. the protocol takes your collateral to pay the winners and then sells your position to someone who can put up fresh collateral.

**Example:** Suppose ETH is trading at 2,000 USD and you want exposure to 1 ETH.

* **Spot:** you pay 2,000 USD and own 1 ETH.
* **Perpetual:** you open a 1 ETH long position and deposit, say, 200 USDC as collateral. Your position is still worth 2,000 USD (its *notional* value), so you are using **10x leverage**.
* If ETH rises to 2,100 USD, you gain **100 USD**, exactly as if you held 1 ETH. That is a 50% return on your 200 USD of collateral.
* If ETH falls to 1,900 USD, you lose **100 USD**, which is taken from your collateral, leaving you with 100 USD.
* If ETH keeps falling and your remaining collateral drops below the protocol's *maintenance margin*, your position is liquidated.

**So how does someone actually profit - or lose money - from a perpetual?**

When they want to close their position, they just sell this long or short perpetual they acquired and will get their profit. If everything is working correctly, they should be able to sell this contract for a profit or loss that corresponds with how they would do if they held the underlying asset. On different protocols "selling" the position can mean selling it on an order book, to an AMM or just getting compensated for their profits to a liquidity pool.  

**How can a protocol ensure that the price of the perpetual will track the underlying asset?**

After all, traders may have a general sentiment, for example that the market is pumping and people expect it will go up a lot, people may be willing to pay more for a long perpetual than what the value should be based on the underlying asset price. To make sure that the price doesn't deviate from the underlying asset price, which is necessary to make this whole thing work, traditional perpetual protocols have a funding rate mechanism that charges periodic fees on the overvalued positions and pays those out to the undervalued positions. This counteracts the market sentiment and drives the price associated with the perpetual to match that of the underlying asset.

**Recap**

* Perpetuals are synthetic assets that allow you to bet on the future price of an underlying asset.
* Perpetuals track the value of the underlying asset through a funding rate mechanism that keeps demand of short and long positions balanced.
* Because perpetuals do not require the purchase or sale of the underlying asset - except for settlement purposes - you can buy a very big position with a small amount of collateral. You just might get wiped out quickly.

Because you are agreeing to buy and sell in the future, perpetuals are considered a futures contract. However, where traditional futures contracts have expiry dates where something has to be exchanged for something, perpetual contracts do not and will last as long as you hold and don't get liquidated.

# DyDx Testnet Trading Competition

DyDx is a perpetuals exchange. The current version (dYdX v4) runs on its own blockchain, the dYdX Chain, built with the Cosmos SDK. It is not on Ethereum or an Ethereum Layer 2: orders are matched by the chain's validators in a fully on-chain order book, and transactions appear on dYdX Chain explorers (such as Mintscan) rather than on Etherscan. (The earlier dYdX v3 was an Ethereum Layer 2 that posted zero-knowledge proofs to Ethereum, but it has since been shut down.)
